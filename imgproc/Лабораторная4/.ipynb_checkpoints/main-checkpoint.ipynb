```python
import cv2
import numpy as np

def clean_noisy_document(image_path, output_path=None):
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    
    if img is None:
        print(f"Ошибка: не удалось загрузить изображение {image_path}")
        return None
    
    # 1. Медианный фильтр для удаления соли-перца и шума (сохраняет края)
    denoised = cv2.medianBlur(img, 5)  # Размер ядра 5x5
    
    # 2. Гауссово размытие для дополнительного сглаживания
    denoised = cv2.GaussianBlur(denoised, (5, 5), 0)
    
    # 3. Нелокальные средства (Non-Local Means) - эффективно для Гауссова шума
    denoised = cv2.fastNlMeansDenoising(denoised, None, h=10, templateWindowSize=7, searchWindowSize=21)
    
    # 4. Адаптивная пороговая бинаризация для восстановления текста
    # Используем OTSU для нахождения оптимального порога
    _, binary = cv2.threshold(denoised, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    
    # 5. Морфологическое открытие для удаления мелких шумовых точек
    kernel = np.ones((2, 2), np.uint8)
    cleaned = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
    
    # 6. Закрытие для соединения разорванных частей букв
    kernel_close = np.ones((3, 3), np.uint8)
    cleaned = cv2.morphologyEx(cleaned, cv2.MORPH_CLOSE, kernel_close)
    
    # 7. Дополнительное удаление мелких объектов (опционально)
    # Удаляем компоненты меньше заданного размера
    num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(cleaned, connectivity=8)
    
    min_area = 50  # Минимальная площадь для текстовых элементов
    for i in range(1, num_labels):
        if stats[i, cv2.CC_STAT_AREA] < min_area:
            cleaned[labels == i] = 0
    
    # Сохранение результата
    if output_path:
        cv2.imwrite(output_path, cleaned)
        print(f"Очищенное изображение сохранено как {output_path}")
    
    return cleaned

def advanced_denoising(image_path, output_path=None):
    """
    Расширенная очистка с несколькими методами для сильного шума (до 90%)
    """
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    
    if img is None:
        return None
    
    # Метод 1: Non-Local Means с более агрессивными параметрами
    denoised_nlm = cv2.fastNlMeansDenoising(img, None, h=15, templateWindowSize=7, searchWindowSize=21)
    
    # Метод 2: Билатеральная фильтрация (сохраняет резкие края)
    denoised_bilateral = cv2.bilateralFilter(img, 9, 75, 75)
    
    # Комбинируем результаты
    denoised = cv2.addWeighted(denoised_nlm, 0.6, denoised_bilateral, 0.4, 0)
    
    # Адаптивная пороговая обработка для неравномерного освещения
    binary = cv2.adaptiveThreshold(denoised, 255, 
                                   cv2.ADAPTIVE_THRESH_GAUSSIAN_C, 
                                   cv2.THRESH_BINARY, 11, 2)
    
    # Удаление шума с помощью морфологии
    kernel = np.ones((3, 3), np.uint8)
    cleaned = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel, iterations=2)
    cleaned = cv2.morphologyEx(cleaned, cv2.MORPH_CLOSE, kernel, iterations=1)
    
    if output_path:
        cv2.imwrite(output_path, cleaned)
    
    return cleaned

# Пример использования
if __name__ == "__main__":
    # Путь к вашему изображению
    input_image = "123.png"
    output_image = "output.png"
    
    # Использование базового метода
    result = clean_noisy_document(input_image, output_image)
    
    # Или используйте расширенный метод для очень сильного шума
    # result = advanced_denoising(input_image, output_image)
    
    if result is not None:
        # Показать результат
        original = cv2.imread(input_image, cv2.IMREAD_GRAYSCALE)
        cv2.imshow("Original", original)
        cv2.imshow("Cleaned", result)
```
