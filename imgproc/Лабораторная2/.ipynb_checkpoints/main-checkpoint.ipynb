```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
image = cv.imread('lenna.png', cv.IMREAD_GRAYSCALE)

#gray_image = cv.cvtColor(image, cv.COLOR_BGR2GRAY)

print(f"Исходное изображение: {image.shape}")
```

```python
hist = cv.calcHist([image], [0], None, [256], [0, 256])

plt.figure(figsize=(14, 3))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.axis('off')
plt.title('Изображение')

plt.subplot(1, 2, 2)
plt.plot(hist, color='black')
plt.title('Гистограмма')
```

```python
hist2, _ = np.histogram(image.flatten(), bins=256, range=[0, 256])
csum = hist.cumsum()
norm = 255 * csum / csum[-1]
lut = np.round(norm).astype(np.uint8)
eq_img = lut[image]
```

```python
hist = cv.calcHist([eq_img], [0], None, [256], [0, 256])

plt.figure(figsize=(14, 3))
plt.subplot(1, 2, 1)
plt.imshow(eq_img, cmap='gray')
plt.axis('off')
plt.title('Изображение')

plt.subplot(1, 2, 2)
plt.plot(hist, color='black')
plt.title('Гистограмма')
```

```python
clahe_filter = cv.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

clahe_image = clahe_filter.apply(image)
```

```python
hist = cv.calcHist([clahe_image], [0], None, [256], [0, 256])

plt.figure(figsize=(14, 3))
plt.subplot(1, 2, 1)
plt.imshow(clahe_image, cmap='gray')
plt.axis('off')
plt.title('Изображение')

plt.subplot(1, 2, 2)
plt.plot(hist, color='black')
plt.title('Гистограмма')
```
